# 11. AdaBoost (Adaptive Boosting)
### Combines Weak Learners Using Adaptive Weights

**AdaBoost** was the first successful boosting algorithm. It trains a sequence of weak learners (typically decision stumps — 1-level decision trees), sequentially assigning higher weights to instances misclassified by earlier learners.

#### Key Concepts:
1. **Sample Weight Update**:
   $$w_i^{(m+1)} = w_i^{(m)} \exp\left(\alpha_m \cdot \mathbb{I}(y_i \neq h_m(x_i))\right)$$
2. **Learner Weight ($\alpha_m$)**:
   $$\alpha_m = \frac{1}{2} \ln\left(\frac{1 - \epsilon_m}{\epsilon_m}\right)$$
   Learners with lower weighted error $\epsilon_m$ receive larger votes.
3. **Final Classification**:
   $$H(x) = \text{sign}\left(\sum_{m=1}^M \alpha_m h_m(x)\right)$$


```python
# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Step 2: Load Data
cancer = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, test_size=0.25, random_state=42, stratify=cancer.target
)
```

```python
# Step 3: Train AdaBoost with Decision Stumps
base_stump = DecisionTreeClassifier(max_depth=1)
ada = AdaBoostClassifier(
    estimator=base_stump,
    n_estimators=100,
    learning_rate=0.5,
    random_state=42
)
ada.fit(X_train, y_train)

y_pred = ada.predict(X_test)
print(f"AdaBoost Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("
Classification Report:
", classification_report(y_test, y_pred, target_names=cancer.target_names))
```

```python
# Step 4: Staged Score Progression
staged_train_acc = [accuracy_score(y_train, p) for p in ada.staged_predict(X_train)]
staged_test_acc = [accuracy_score(y_test, p) for p in ada.staged_predict(X_test)]

plt.figure(figsize=(9, 5))
plt.plot(range(1, len(staged_train_acc) + 1), staged_train_acc, label='Train Accuracy', color='blue')
plt.plot(range(1, len(staged_test_acc) + 1), staged_test_acc, label='Test Accuracy', color='orange')
plt.xlabel("Number of Estimators (Boosting Iterations)")
plt.ylabel("Accuracy")
plt.title("AdaBoost Performance vs Number of Iterations")
plt.legend()
plt.grid(True)
plt.show()
```

### Summary
- **Strengths**: Simple, mathematically elegant, converts simple weak learners (stumps) into high-accuracy classifiers without requiring deep trees.
- **Weaknesses**: Sensitive to noisy data and outliers because misclassified outlier points receive exponentially increasing weights.
