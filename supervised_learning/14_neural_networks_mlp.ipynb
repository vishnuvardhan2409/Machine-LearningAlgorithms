# 14. Neural Networks: Multi-Layer Perceptron (MLP)
### Learns Nonlinear Patterns

A **Multi-Layer Perceptron (MLP)** is a feedforward artificial neural network consisting of an input layer, one or more hidden layers with non-linear activation functions, and an output layer.

#### Key Concepts:
1. **Forward Propagation**:
   $$a^{(l)} = g(W^{(l)} a^{(l-1)} + b^{(l)})$$
   where $g(\cdot)$ is a non-linear activation (e.g. ReLU, Tanh, Sigmoid).
2. **Backpropagation**:
   Applies chain rule of calculus to compute gradient of the loss with respect to all layer weights:
   $$\frac{\partial \mathcal{L}}{\partial W^{(l)}}$$
3. **Optimization**:
   Optimizers like **Adam** or **SGD with momentum** update weights iteratively to minimize loss.
4. **Regularization**:
   Weight decay ($L_2$ regularization `alpha`) and early stopping prevent overfitting.


```python
# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
```

```python
# Step 2: Load and Scale Breast Cancer Dataset
cancer = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, test_size=0.25, random_state=42, stratify=cancer.target
)

# Standard scaling is ESSENTIAL for neural networks
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Step 3: Train MLP Classifier
mlp = MLPClassifier(
    hidden_layer_sizes=(64, 32),
    activation='relu',
    solver='adam',
    alpha=0.01,           # L2 regularization
    max_iter=500,
    random_state=42,
    early_stopping=True,
    n_iter_no_change=15
)
mlp.fit(X_train_scaled, y_train)

y_pred = mlp.predict(X_test_scaled)
print(f"MLP Test Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print(f"Total Iterations to Converge: {mlp.n_iter_}")
print("
Classification Report:
", classification_report(y_test, y_pred, target_names=cancer.target_names))
```

```python
# Step 4: Visualizing Training Loss Curve
plt.figure(figsize=(8, 4))
plt.plot(mlp.loss_curve_, color='crimson', lw=2)
plt.title("MLP Training Loss Curve")
plt.xlabel("Epochs / Iterations")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```

```python
# Step 5: Confusion Matrix
fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay.from_estimator(mlp, X_test_scaled, y_test, display_labels=cancer.target_names, cmap='Blues', ax=ax)
plt.title("MLP Confusion Matrix")
plt.show()
```

### Summary
- **Strengths**: Universal approximator capable of learning highly complex non-linear relationships, scalable to complex multimodal architectures.
- **Weaknesses**: Computationally demanding, requires careful scaling and hyperparameter tuning, behaves largely as a black box.
