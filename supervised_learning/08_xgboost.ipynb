# 8. XGBoost (Extreme Gradient Boosting)
### Optimized Gradient Boosting

**XGBoost** is an optimized, scalable, distributed gradient boosted decision tree library designed for efficiency, flexibility, and portability.

#### Key Innovations:
1. **Second-Order Taylor Approximation**: Uses both 1st derivatives (gradients $g_i$) and 2nd derivatives (hessians $h_i$) of the loss function.
2. **Regularized Objective**: Includes both $L_1$ (Lasso, $\alpha$) and $L_2$ (Ridge, $\lambda$) regularization on leaf weights to avoid overfitting:
   $$\mathcal{L} = \sum_i l(y_i, \hat{y}_i) + \sum_k \left(\gamma T_k + \frac{1}{2} \lambda \sum_j w_{kj}^2\right)$$
3. **Sparsity-aware Split Finding**: Efficient handling of missing values.
4. **Column Subsampling and Block Structure**: Parallel tree construction and cache-aware access.


```python
# Step 1: Install XGBoost if needed & Import Libraries
# !pip install xgboost

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

try:
    import xgboost as xgb
    print(f"XGBoost version: {xgb.__version__}")
except ImportError:
    print("XGBoost is not installed in this environment. Please run: pip install xgboost")
```

```python
# Step 2: Load Dataset
cancer = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, test_size=0.25, random_state=42, stratify=cancer.target
)
```

```python
# Step 3: Train XGBoost Classifier
try:
    xgb_model = xgb.XGBClassifier(
        n_estimators=100,
        learning_rate=0.08,
        max_depth=4,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        reg_alpha=0.1,
        random_state=42,
        eval_metric='logloss'
    )
    xgb_model.fit(X_train, y_train, eval_set=[(X_test, y_test)], verbose=False)

    y_pred = xgb_model.predict(X_test)
    print(f"XGBoost Test Accuracy: {accuracy_score(y_test, y_pred):.4f}")
    print("
Classification Report:
", classification_report(y_test, y_pred, target_names=cancer.target_names))
except NameError:
    print("Install xgboost to run this cell.")
```

```python
# Step 4: Plot Feature Importance
try:
    plt.figure(figsize=(10, 6))
    xgb.plot_importance(xgb_model, max_num_features=10, height=0.6, color='steelblue')
    plt.title("XGBoost Top 10 Feature Importances (Weight/Frequency)")
    plt.show()
except Exception as e:
    print(e)
```

### Summary
- **Strengths**: Lightning fast execution, built-in regularization, built-in cross-validation and early stopping, dominant in competitive machine learning (Kaggle).
- **Weaknesses**: Many hyperparameters to tune, higher memory usage than lightgbm.
