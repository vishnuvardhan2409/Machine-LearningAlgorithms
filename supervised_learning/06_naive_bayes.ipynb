# 6. Naive Bayes
### Probabilistic Classification

**Naive Bayes** classifiers are a family of probabilistic algorithms based on **Bayes' Theorem** with the strong ("naive") assumption of conditional independence between every pair of features given the class label.

#### Key Concepts:
1. **Bayes' Theorem**:
   $$P(Y | X) = \frac{P(X | Y) P(Y)}{P(X)}$$
2. **Conditional Independence Assumption**:
   $$P(X_1, X_2, \dots, X_n | Y) = \prod_{i=1}^n P(X_i | Y)$$
3. **Variants**:
   - **GaussianNB**: Assumes continuous features follow a Gaussian distribution.
   - **MultinomialNB**: Ideal for discrete counts (e.g. text / NLP word frequencies).
   - **BernoulliNB**: Ideal for binary boolean features.


```python
# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
```

```python
# Step 2: Load Wine Dataset
wine = load_wine()
X = wine.data
y = wine.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python
# Step 3: Train Gaussian Naive Bayes
gnb = GaussianNB()
gnb.fit(X_train, y_train)

y_pred = gnb.predict(X_test)
y_probs = gnb.predict_proba(X_test)

print(f"Gaussian Naive Bayes Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("
Classification Report:
", classification_report(y_test, y_pred, target_names=wine.target_names))
```

```python
# Step 4: Class Priors and Learned Parameters
print("Prior Probabilities of Classes:")
for cls, prior in zip(wine.target_names, gnb.class_prior_):
    print(f"  {cls}: {prior:.4f}")

# Display predicted class probabilities for first 5 test samples
prob_df = pd.DataFrame(y_probs[:5], columns=wine.target_names)
prob_df['Predicted'] = [wine.target_names[i] for i in y_pred[:5]]
prob_df['Actual'] = [wine.target_names[i] for i in y_test[:5]]
print("
Predicted Posterior Probabilities for first 5 samples:")
print(prob_df)
```

```python
# Step 5: Confusion Matrix
fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay.from_estimator(gnb, X_test, y_test, display_labels=wine.target_names, cmap='Blues', ax=ax)
plt.title("Naive Bayes Confusion Matrix")
plt.show()
```

### Summary
- **Strengths**: Extremely fast to train and predict, works well with high-dimensional data (e.g., spam filtering, text classification), requires little training data.
- **Weaknesses**: Naive independence assumption rarely holds strictly in reality; correlated features can distort probability estimates.
