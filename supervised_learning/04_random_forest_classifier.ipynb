# 4. Random Forest Classifier
### Ensemble of Decision Trees

A **Random Forest** is an ensemble method that builds multiple de-correlated decision trees using **Bootstrap Aggregating (Bagging)** and random feature selection, combining their predictions via majority voting.

#### Key Concepts:
1. **Bagging (Bootstrap Sampling)**: Each tree is trained on a random sample drawn with replacement from the training set.
2. **Feature Randomness**: At each split, only a random subset of features (typically $\sqrt{p}$) is considered, decorrelating individual trees.
3. **Variance Reduction**: Averaging predictions across trees drastically reduces model variance without increasing bias.
4. **Out-of-Bag (OOB) Score**: Uses unsampled data points to validate individual trees without a separate validation split.


```python
# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
```

```python
# Step 2: Load and Split Wine Dataset
wine = load_wine()
X = wine.data
y = wine.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
```

```python
# Step 3: Train Random Forest with Out-of-Bag (OOB) Evaluation
rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    max_features='sqrt',
    oob_score=True,
    random_state=42
)
rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)
print(f"Out-of-Bag (OOB) Accuracy: {rf.oob_score_:.4f}")
print(f"Test Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("
Classification Report:
", classification_report(y_test, y_pred, target_names=wine.target_names))
```

```python
# Step 4: Feature Importance Analysis
feature_imp = pd.Series(rf.feature_importances_, index=wine.feature_names).sort_values(ascending=False)

plt.figure(figsize=(10, 5))
feature_imp.plot(kind='bar', color='forestgreen')
plt.title("Random Forest Mean Decrease in Impurity (Feature Importances)")
plt.ylabel("Importance Score")
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Step 5: Effect of Number of Trees (n_estimators)
estimator_range = [5, 10, 20, 50, 100, 150, 200]
accuracies = []

for n in estimator_range:
    clf = RandomForestClassifier(n_estimators=n, random_state=42)
    clf.fit(X_train, y_train)
    accuracies.append(clf.score(X_test, y_test))

plt.figure(figsize=(8, 4))
plt.plot(estimator_range, accuracies, marker='o', color='darkgreen')
plt.xlabel("Number of Estimators (Trees)")
plt.ylabel("Test Accuracy")
plt.title("Random Forest: Accuracy vs Number of Trees")
plt.grid(True)
plt.show()
```

### Summary
- **Strengths**: Robust against overfitting, handles high-dimensional data, delivers high accuracy out of the box, provides feature importances and OOB score.
- **Weaknesses**: Slower inference and larger memory footprint than single trees, less interpretable than a single decision tree.
