# 7. Gradient Boosting
### Sequentially Improves Weak Learners

**Gradient Boosting** is an ensemble technique that builds trees sequentially. Each new tree fits to the **pseudo-residuals (negative gradient)** of the loss function associated with the whole ensemble.

#### Key Concepts:
1. **Additive Modeling**:
   $$F_m(x) = F_{m-1}(x) + \eta \cdot h_m(x)$$
   where $\eta$ is the learning rate (shrinkage) and $h_m(x)$ is the new weak learner.
2. **Gradient Descent in Function Space**: Each tree fits the direction of steepest descent for the loss function (e.g., Deviance/Log-loss).
3. **Hyperparameters**:
   - `n_estimators`: Total number of boosting stages.
   - `learning_rate`: Shrinks contribution of each tree to prevent overfitting.
   - `subsample`: Fraction of samples used for fitting trees (Stochastic Gradient Boosting).


```python
# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Step 2: Load Dataset
cancer = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, test_size=0.25, random_state=42, stratify=cancer.target
)
```

```python
# Step 3: Train Gradient Boosting Classifier
gb = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    subsample=0.8,
    random_state=42
)
gb.fit(X_train, y_train)

y_pred = gb.predict(X_test)
print(f"Gradient Boosting Test Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("
Classification Report:
", classification_report(y_test, y_pred, target_names=cancer.target_names))
```

```python
# Step 4: Staged Deviance / Cumulative Accuracy over Iterations
train_deviance = [gb.loss_(y_train, y_pred_staged) for y_pred_staged in gb.staged_decision_function(X_train)]
test_deviance = [gb.loss_(y_test, y_pred_staged) for y_pred_staged in gb.staged_decision_function(X_test)]

plt.figure(figsize=(9, 5))
plt.plot(np.arange(1, len(train_deviance) + 1), train_deviance, label='Train Loss (Deviance)', color='blue')
plt.plot(np.arange(1, len(test_deviance) + 1), test_deviance, label='Test Loss (Deviance)', color='red')
plt.xlabel("Boosting Iterations (n_estimators)")
plt.ylabel("Deviance")
plt.title("Gradient Boosting Training vs Test Loss per Iteration")
plt.legend()
plt.grid(True)
plt.show()
```

### Summary
- **Strengths**: Very high predictive accuracy, flexible loss functions, handles mixed feature types naturally.
- **Weaknesses**: Prone to overfitting if `n_estimators` is too high without tuning `learning_rate`, slower to train sequentially than parallel methods.
