# 9. LightGBM
### Efficient Gradient Boosting

**LightGBM** (Light Gradient Boosting Machine) developed by Microsoft is known for fast training speed and low memory usage on large tabular datasets.

#### Key Innovations:
1. **Leaf-wise (Best-first) Tree Growth**: Grows trees leaf-wise rather than level-wise, reducing loss much faster per split.
2. **GOSS (Gradient-based One-Side Sampling)**: Keeps data instances with large gradients and randomly drops instances with small gradients.
3. **EFB (Exclusive Feature Bundling)**: Bundles mutually exclusive sparse features to reduce dimension count.
4. **Histogram-based Algorithm**: Buckets continuous feature values into discrete bins, speeding up split finding significantly.


```python
# Step 1: Install LightGBM if needed & Import Libraries
# !pip install lightgbm

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

try:
    import lightgbm as lgb
    print(f"LightGBM version: {lgb.__version__}")
except ImportError:
    print("LightGBM is not installed. Please run: pip install lightgbm")
```

```python
# Step 2: Prepare Data
cancer = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, test_size=0.25, random_state=42, stratify=cancer.target
)
```

```python
# Step 3: Train LightGBM Classifier
try:
    lgbm_model = lgb.LGBMClassifier(
        n_estimators=100,
        learning_rate=0.05,
        num_leaves=31,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    )
    lgbm_model.fit(X_train, y_train)

    y_pred = lgbm_model.predict(X_test)
    print(f"LightGBM Test Accuracy: {accuracy_score(y_test, y_pred):.4f}")
    print("
Classification Report:
", classification_report(y_test, y_pred, target_names=cancer.target_names))
except NameError:
    print("Please install lightgbm to execute.")
```

```python
# Step 4: LightGBM Feature Importance
try:
    plt.figure(figsize=(10, 5))
    lgb.plot_importance(lgbm_model, max_num_features=10, color='coral')
    plt.title("LightGBM Top 10 Feature Importances")
    plt.show()
except Exception as e:
    print(e)
```

### Summary
- **Strengths**: Faster training speed and higher efficiency than standard GBDT, lower memory usage, handles large-scale datasets with millions of rows.
- **Weaknesses**: Can overfit small datasets if `num_leaves` and `max_depth` are not properly constrained.
