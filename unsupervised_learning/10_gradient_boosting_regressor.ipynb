# 10. Gradient Boosting Regressor
### Sequential Error Correction

**Gradient Boosting Regressor** iteratively constructs an ensemble of shallow regression trees, where each subsequent tree fits to the **negative gradient (residuals)** of the squared error loss function of previous predictions.

#### Key Concepts:
1. **Iterative Residual Fitting**:
   $$r_{im} = - \left[ \frac{\partial L(y_i, F(x_i))}{\partial F(x_i)} \right]_{F(x) = F_{m-1}(x)}$$
   For squared error loss, the negative gradient is simply the residual: $y_i - F_{m-1}(x_i)$.
2. **Shrinkage ($\\eta$)**:
   $$F_m(x) = F_{m-1}(x) + \eta \cdot h_m(x)$$
   Controls learning rate to avoid premature overfitting.


```python
# Step 1: Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Step 2: Load Data
housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(
    housing.data[:2000], housing.target[:2000], test_size=0.25, random_state=42
)
```

```python
# Step 3: Train Gradient Boosting Regressor
gbr = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=4,
    subsample=0.8,
    random_state=42
)
gbr.fit(X_train, y_train)

y_pred = gbr.predict(X_test)
print(f"Test R² Score: {r2_score(y_test, y_pred):.4f}")
print(f"Test RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
```

```python
# Step 4: Staged Loss / Deviances Progression
train_loss = [gbr.loss_(y_train, y_pred_s) for y_pred_s in gbr.staged_predict(X_train)]
test_loss = [gbr.loss_(y_test, y_pred_s) for y_pred_s in gbr.staged_predict(X_test)]

plt.figure(figsize=(9, 5))
plt.plot(train_loss, label='Train MSE Loss', color='blue')
plt.plot(test_loss, label='Test MSE Loss', color='red')
plt.xlabel("Boosting Iterations (Trees)")
plt.ylabel("Loss (MSE)")
plt.title("Gradient Boosting Regressor: Training vs Test Loss per Iteration")
plt.legend()
plt.grid(True)
plt.show()
```

### Summary
- **Strengths**: High predictive accuracy on heterogeneous tabular data, robust loss optimization.
- **Weaknesses**: Sequential training makes it slower than parallel bagging; hyperparameter tuning (learning rate and trees) is required to avoid overfitting.
