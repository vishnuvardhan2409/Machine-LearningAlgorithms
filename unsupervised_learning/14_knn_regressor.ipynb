# 14. K-Nearest Neighbors (KNN) Regressor
### Predicts Based on Nearby Observations

**KNN Regressor** estimates the numerical target of a query point by averaging (uniform or inverse-distance weighted) the target values of its $k$ closest neighbors in feature space.

#### Key Concepts:
1. **Prediction Rule**:
   - Uniform weights: $\hat{y} = \frac{1}{k} \sum_{i \in N_k(x)} y_i$
   - Distance weights: $\hat{y} = \frac{\sum_{i \in N_k(x)} w_i y_i}{\sum_{i \in N_k(x)} w_i}$ where $w_i = 1 / d(x, x_i)$.
2. **Scale Sensitivity**:
   Mandatory feature standardization so that distance calculations are not dominated by features with large absolute scales.


```python
# Step 1: Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Step 2: Load and Scale Diabetes Data
diabetes = load_diabetes()
X_train, X_test, y_train, y_test = train_test_split(
    diabetes.data, diabetes.target, test_size=0.25, random_state=42
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Step 3: Train KNN Regressor
k = 7
knn_reg = KNeighborsRegressor(n_neighbors=k, weights='distance')
knn_reg.fit(X_train_scaled, y_train)

y_pred = knn_reg.predict(X_test_scaled)
print(f"KNN (k={k}) Test R²: {r2_score(y_test, y_pred):.4f}")
print(f"KNN Test RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
```

```python
# Step 4: Finding the Optimal K Value
k_values = range(1, 30)
rmse_scores = []

for cur_k in k_values:
    reg = KNeighborsRegressor(n_neighbors=cur_k, weights='distance')
    reg.fit(X_train_scaled, y_train)
    rmse_scores.append(np.sqrt(mean_squared_error(y_test, reg.predict(X_test_scaled))))

plt.figure(figsize=(9, 5))
plt.plot(k_values, rmse_scores, marker='o', color='darkorange')
plt.xlabel("Value of K")
plt.ylabel("Test RMSE")
plt.title("KNN Regressor: RMSE vs Value of K")
plt.grid(True)
plt.show()
```

### Summary
- **Strengths**: Non-parametric, zero training time, naturally adapts to local non-linear manifolds.
- **Weaknesses**: Slow inference time $O(N \cdot D)$, suffers heavily from the curse of dimensionality, cannot extrapolate beyond training bounds.
