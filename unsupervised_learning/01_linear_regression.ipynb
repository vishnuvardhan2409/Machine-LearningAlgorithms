# 1. Simple Linear Regression
### Predict Continuous Values

**Simple Linear Regression** models the linear relationship between a single independent explanatory variable ($X$) and a continuous target variable ($y$).

#### Key Concepts:
1. **Hypothesis / Model Equation**:
   $$\hat{y} = w_0 + w_1 x$$
   where $w_0$ is the intercept and $w_1$ is the slope coefficient.
2. **Cost Function (Ordinary Least Squares - OLS)**:
   $$J(w) = \frac{1}{2m} \sum_{i=1}^m (y_i - \hat{y}_i)^2$$
   Minimizes the sum of squared differences (residuals) between observed values and predictions.
3. **Closed-Form Normal Equation**:
   $$w = (X^T X)^{-1} X^T y$$


```python
# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
# Step 2: Generate Synthetic 1D Regression Data
np.random.seed(42)
X = 2 * np.random.rand(150, 1)
# y = 4 + 3*X + Gaussian noise
y = 4 + 3 * X.squeeze() + np.random.randn(150) * 0.8

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
# Step 3: Fit Linear Regression Model
lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred = lr.predict(X_test)

print(f"Intercept (w0): {lr.intercept_:.4f}")
print(f"Slope (w1): {lr.coef_[0]:.4f}")
print(f"R² Score: {r2_score(y_test, y_pred):.4f}")
print(f"Mean Absolute Error (MAE): {mean_absolute_error(y_test, y_pred):.4f}")
print(f"Root Mean Squared Error (RMSE): {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
```

```python
# Step 4: Visualize the Regression Line and Residuals
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Plot regression line
ax1.scatter(X_train, y_train, color='blue', alpha=0.6, label='Train data')
ax1.scatter(X_test, y_test, color='green', alpha=0.8, label='Test data')
X_line = np.linspace(0, 2, 100).reshape(-1, 1)
ax1.plot(X_line, lr.predict(X_line), color='red', lw=2, label='Fitted Line')
ax1.set_xlabel("X")
ax1.set_ylabel("y")
ax1.set_title("Simple Linear Regression Fit")
ax1.legend()
ax1.grid(True)

# Residual plot
residuals = y_test - y_pred
ax2.scatter(y_pred, residuals, color='purple', alpha=0.7)
ax2.axhline(0, color='black', linestyle='--')
ax2.set_xlabel("Predicted Values")
ax2.set_ylabel("Residuals")
ax2.set_title("Residual Plot")
ax2.grid(True)

plt.show()
```

### Summary
- **Strengths**: Highly interpretable, exact analytical solution via OLS, fast computation.
- **Weaknesses**: Assumes strict linearity between feature and target; vulnerable to outliers and homoscedasticity violations.
