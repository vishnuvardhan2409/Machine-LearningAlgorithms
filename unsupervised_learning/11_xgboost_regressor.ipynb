# 11. XGBoost Regressor
### Optimized Gradient Boosting

**XGBoost Regressor** is an optimized gradient boosting framework implementing 2nd-order Taylor expansions of the objective, $L_1$ and $L_2$ leaf regularization, and cache-aware hardware acceleration.

#### Key Features:
- Exact and approximate histogram greedy split finding.
- Built-in handling of missing values.
- Column subsampling at tree, level, and split scales.
- Early stopping based on validation metrics.


```python
# Step 1: Install XGBoost if needed & Import Libraries
# !pip install xgboost

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

try:
    import xgboost as xgb
    print("XGBoost version:", xgb.__version__)
except ImportError:
    print("XGBoost not installed. Run: pip install xgboost")
```

```python
# Step 2: Load Data
housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(
    housing.data[:2000], housing.target[:2000], test_size=0.25, random_state=42
)
```

```python
# Step 3: Train XGBoost Regressor
try:
    xgb_reg = xgb.XGBRegressor(
        n_estimators=100,
        learning_rate=0.08,
        max_depth=5,
        reg_lambda=1.5,
        reg_alpha=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    )
    xgb_reg.fit(X_train, y_train, eval_set=[(X_test, y_test)], verbose=False)

    y_pred = xgb_reg.predict(X_test)
    print(f"XGBoost Test R²: {r2_score(y_test, y_pred):.4f}")
    print(f"XGBoost Test RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
except NameError:
    print("XGBoost not available.")
```

```python
# Step 4: Plot Top Feature Importances
try:
    plt.figure(figsize=(9, 5))
    xgb.plot_importance(xgb_reg, max_num_features=8, height=0.6, color='steelblue')
    plt.title("XGBoost Regressor: Feature Importance")
    plt.show()
except Exception as e:
    print(e)
```

### Summary
- **Strengths**: Superb performance on tabular regression, highly regularized to prevent overfitting, industry standard for competitive data science.
- **Weaknesses**: Hyperparameter tuning requires care across depth, learning rate, and regularizations.
