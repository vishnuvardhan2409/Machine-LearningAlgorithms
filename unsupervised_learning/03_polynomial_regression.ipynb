# 3. Polynomial Regression
### Models Nonlinear Relationships

**Polynomial Regression** captures non-linear relationships by creating polynomial feature combinations ($x^2, x^3, \dots, x^d$) while still remaining a linear model in parameters $w$.

#### Key Concepts:
1. **Model Formulation**:
   $$\hat{y} = w_0 + w_1 x + w_2 x^2 + \dots + w_d x^d$$
2. **Feature Transformation**:
   Uses `PolynomialFeatures(degree=d)` to project inputs into higher-degree feature space.
3. **Bias-Variance Tradeoff**:
   - Degree too low ($d=1$): High bias (Underfitting).
   - Degree too high ($d \ge 10$): High variance (Overfitting, oscillations at boundaries — Runge's phenomenon).


```python
# Step 1: Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Step 2: Generate Non-linear Quadratic/Cubic Synthetic Data
np.random.seed(42)
m = 100
X = 6 * np.random.rand(m, 1) - 3
y = 0.5 * X.squeeze()**2 + X.squeeze() + 2 + np.random.randn(m) * 0.8

# Test points for smooth curve plotting
X_seq = np.linspace(-3, 3, 200).reshape(-1, 1)
```

```python
# Step 3: Compare Degrees 1, 2, and 10
degrees = [1, 2, 10]
plt.figure(figsize=(12, 5))
plt.scatter(X, y, color='black', alpha=0.6, label='Data points')

colors = ['orange', 'green', 'red']
for deg, color in zip(degrees, colors):
    poly_reg = make_pipeline(PolynomialFeatures(degree=deg), LinearRegression())
    poly_reg.fit(X, y)
    y_plot = poly_reg.predict(X_seq)
    r2 = r2_score(y, poly_reg.predict(X))
    plt.plot(X_seq, y_plot, color=color, lw=2, label=f"Degree {deg} (R² = {r2:.3f})")

plt.xlabel("X")
plt.ylabel("y")
plt.title("Polynomial Regression: Underfitting vs Optimal vs Overfitting")
plt.legend()
plt.ylim(-2, 12)
plt.grid(True)
plt.show()
```

### Summary
- **Strengths**: Models curved and non-linear patterns within the framework of linear regression.
- **Weaknesses**: Susceptible to severe overfitting with high degrees; extreme predictions when extrapolating outside the training range.
