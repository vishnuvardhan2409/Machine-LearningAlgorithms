# 4. Ridge Regression (L2 Regularization)
### Linear Regression with L2 Regularization (Tikhonov)

**Ridge Regression** addresses multicollinearity and overfitting in OLS by adding an $L_2$ penalty proportional to the square of coefficient magnitudes.

#### Key Concepts:
1. **Regularized Cost Function**:
   $$J(w) = \frac{1}{2m} \sum_{i=1}^m (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^p w_j^2$$
2. **Shrinkage Effect**:
   Shrinks coefficients towards zero, reducing variance without zeroing them out completely.
3. **Closed-Form Solution**:
   $$w = (X^T X + \alpha I)^{-1} X^T y$$
   The $\alpha I$ term ensures invertibility even when $X^T X$ is singular (high collinearity).


```python
# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge, RidgeCV
from sklearn.metrics import r2_score, mean_squared_error
```

```python
# Step 2: Load Data and Standardize Features
diabetes = load_diabetes()
X = diabetes.data
y = diabetes.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Regularization is scale-sensitive!
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Step 3: Cross-Validated Ridge Regression (Find Optimal Alpha)
alphas = np.logspace(-3, 3, 50)
ridge_cv = RidgeCV(alphas=alphas, cv=5)
ridge_cv.fit(X_train_scaled, y_train)

y_pred = ridge_cv.predict(X_test_scaled)
print(f"Optimal Alpha: {ridge_cv.alpha_:.4f}")
print(f"Ridge Test R² Score: {r2_score(y_test, y_pred):.4f}")
print(f"Ridge Test RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
```

```python
# Step 4: Coefficient Trace (Regularization Path)
coefs = []
for a in alphas:
    model = Ridge(alpha=a)
    model.fit(X_train_scaled, y_train)
    coefs.append(model.coef_)

plt.figure(figsize=(10, 5))
plt.plot(alphas, coefs)
plt.xscale('log')
plt.xlabel("Alpha (Regularization Strength)")
plt.ylabel("Coefficients")
plt.title("Ridge Coefficients as a Function of Regularization Strength (Alpha)")
plt.axvline(ridge_cv.alpha_, color='k', linestyle='--', label=f'Best Alpha: {ridge_cv.alpha_:.2f}')
plt.legend()
plt.grid(True)
plt.show()
```

### Summary
- **Strengths**: Effectively tackles multicollinearity, keeps all features while taming large unstable coefficients, has a guaranteed closed-form solution.
- **Weaknesses**: Does not perform feature selection (does not set coefficients strictly to zero).
