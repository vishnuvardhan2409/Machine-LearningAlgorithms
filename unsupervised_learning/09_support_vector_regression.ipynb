# 9. Support Vector Regression (SVR)
### Regression using Support Vectors and Epsilon-Insensitive Loss

**Support Vector Regression (SVR)** finds a function that has at most $\epsilon$ deviation from actual target values for all training data while being as flat as possible.

#### Key Concepts:
1. **$\epsilon$-Insensitive Loss Function**:
   $$L_\epsilon(y, \hat{y}) = \max(0, |y - \hat{y}| - \epsilon)$$
   Errors smaller than $\epsilon$ are penalized with zero cost (the "$\epsilon$-tube").
2. **Support Vectors**:
   Only data points lying on or outside the $\epsilon$-tube boundaries act as support vectors that define the prediction function.
3. **Kernels**:
   RBF, Linear, or Polynomial kernels allow fitting non-linear regression boundaries in high-dimensional feature spaces.


```python
# Step 1: Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Step 2: Generate Non-linear 1D Regression Data
np.random.seed(42)
X = np.sort(5 * np.random.rand(60, 1), axis=0)
y = np.sin(X).ravel() + np.random.normal(0, 0.15, 60)

scaler_x = StandardScaler()
scaler_y = StandardScaler()

X_scaled = scaler_x.fit_transform(X)
y_scaled = scaler_y.fit_transform(y.reshape(-1, 1)).ravel()

X_test_plot = np.linspace(0, 5, 200).reshape(-1, 1)
X_test_scaled = scaler_x.transform(X_test_plot)
```

```python
# Step 3: Fit SVR with RBF Kernel
epsilon = 0.2
svr = SVR(kernel='rbf', C=10.0, epsilon=epsilon, gamma='scale')
svr.fit(X_scaled, y_scaled)

y_pred_scaled = svr.predict(X_test_scaled)
y_pred_plot = scaler_y.inverse_transform(y_pred_scaled.reshape(-1, 1)).ravel()

print(f"Number of Support Vectors: {len(svr.support_)}")
print(f"Percentage of samples as SVs: {len(svr.support_) / len(X) * 100:.1f}%")
```

```python
# Step 4: Visualize SVR Prediction Tube and Support Vectors
plt.figure(figsize=(10, 5))
plt.scatter(X, y, color='darkorange', label='Data points', edgecolors='k')
plt.scatter(X[svr.support_], y[svr.support_], s=120, facecolors='none', edgecolors='red', lw=1.5, label='Support Vectors')
plt.plot(X_test_plot, y_pred_plot, color='navy', lw=2, label='SVR (RBF)')

# Invert epsilon bounds for visualization
y_upper = scaler_y.inverse_transform((y_pred_scaled + epsilon).reshape(-1, 1)).ravel()
y_lower = scaler_y.inverse_transform((y_pred_scaled - epsilon).reshape(-1, 1)).ravel()
plt.fill_between(X_test_plot.ravel(), y_lower, y_upper, color='navy', alpha=0.15, label='ε-tube')

plt.xlabel("X")
plt.ylabel("y")
plt.title("Support Vector Regression: RBF Kernel with ε-Insensitive Tube")
plt.legend()
plt.grid(True)
plt.show()
```

### Summary
- **Strengths**: Robust to outliers outside the margin, effective in high-dimensional feature spaces, non-linear modeling via kernels.
- **Weaknesses**: Highly sensitive to feature scaling; quadratic time complexity $O(N^2)$ to $O(N^3)$ makes it slow on very large datasets.
