# 7. Decision Tree Regressor
### Rule-based Numerical Prediction

A **Decision Tree Regressor** recursively splits the feature space into non-overlapping axis-aligned rectangular partitions, predicting the constant mean target value of training samples inside each terminal leaf node.

#### Key Concepts:
1. **Splitting Criterion**:
   Minimizes Variance / Mean Squared Error (MSE) within candidate splits:
   $$\text{MSE} = \frac{1}{N} \sum_{i=1}^N (y_i - \bar{y})^2$$
2. **Piecewise Constant Approximations**:
   The output is a step-function across partition boundaries.
3. **Regularization Hyperparameters**:
   - `max_depth`: Limits tree depth to prevent memorizing training noise.
   - `min_samples_split`: Minimum instances required to attempt a split.
   - `min_samples_leaf`: Minimum instances required to exist at a leaf.


```python
# Step 1: Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Step 2: Generate 1D Non-linear Continuous Data (Sine Wave with Noise)
np.random.seed(42)
X = np.sort(5 * np.random.rand(80, 1), axis=0)
y = np.sin(X).ravel() + np.random.normal(0, 0.1, X.shape[0])

X_test = np.linspace(0, 5, 500).reshape(-1, 1)
```

```python
# Step 3: Compare Trees with Different Depths (depth=2 vs depth=5)
dt_shallow = DecisionTreeRegressor(max_depth=2, random_state=42).fit(X, y)
dt_deep = DecisionTreeRegressor(max_depth=5, random_state=42).fit(X, y)

y_pred_shallow = dt_shallow.predict(X_test)
y_pred_deep = dt_deep.predict(X_test)

plt.figure(figsize=(12, 5))
plt.scatter(X, y, s=30, edgecolor="black", c="darkorange", label="Data points")
plt.plot(X_test, y_pred_shallow, color="cornflowerblue", label="max_depth=2", linewidth=2)
plt.plot(X_test, y_pred_deep, color="yellowgreen", label="max_depth=5 (Overfitting)", linewidth=2)
plt.xlabel("Feature X")
plt.ylabel("Target y")
plt.title("Decision Tree Regressor: Step Function Fit")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Step 4: Visualize Tree Architecture (Shallow Tree)
plt.figure(figsize=(10, 6))
plot_tree(dt_shallow, filled=True, rounded=True, feature_names=['X'])
plt.title("Decision Tree Regressor Architecture (max_depth=2)")
plt.show()
```

### Summary
- **Strengths**: Highly interpretable, requires zero feature normalization, captures non-linear relationships.
- **Weaknesses**: Stepwise predictions (non-smooth), cannot extrapolate outside training range, prone to high variance.
