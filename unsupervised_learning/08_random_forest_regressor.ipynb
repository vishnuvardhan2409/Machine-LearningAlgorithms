# 8. Random Forest Regressor
### Ensemble-based Regression

A **Random Forest Regressor** trains an ensemble of de-correlated decision tree regressors using bootstrap aggregation (bagging) and feature subsampling, producing the final numerical prediction by averaging the predictions of all individual trees.

#### Key Concepts:
1. **Prediction Averaging**:
   $$\hat{y} = \frac{1}{B} \sum_{b=1}^B T_b(x)$$
2. **Variance Reduction**:
   Ensembling uncorrelated trees averages out individual tree errors, drastically reducing variance without increasing bias.
3. **Out-of-Bag (OOB) Evaluation**:
   Estimates generalization performance using unselected bootstrap samples without needing an explicit cross-validation set.


```python
# Step 1: Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Step 2: Load California Housing Dataset
housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(
    housing.data[:2000], housing.target[:2000], test_size=0.25, random_state=42
)
```

```python
# Step 3: Train Random Forest Regressor
rf = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    max_features='sqrt',
    oob_score=True,
    random_state=42
)
rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)
print(f"OOB R² Score: {rf.oob_score_:.4f}")
print(f"Test R² Score: {r2_score(y_test, y_pred):.4f}")
print(f"Test RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
```

```python
# Step 4: Feature Importances
importances = pd.Series(rf.feature_importances_, index=housing.feature_names).sort_values(ascending=False)

plt.figure(figsize=(9, 4))
importances.plot(kind='barh', color='forestgreen')
plt.title("Random Forest Regressor: Feature Importances")
plt.xlabel("MDI Importance")
plt.gca().invert_yaxis()
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

### Summary
- **Strengths**: Robust to outliers and noise, avoids overfitting compared to single decision trees, provides out-of-bag validation.
- **Weaknesses**: Computationally expensive for large tree counts, unable to extrapolate outside bounds of observed training target values.
