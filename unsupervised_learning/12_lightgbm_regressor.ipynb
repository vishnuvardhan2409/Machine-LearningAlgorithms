# 12. LightGBM Regressor
### Fast Gradient Boosting

**LightGBM Regressor** is engineered by Microsoft for exceptional training speed and low memory usage using **histogram-based binning** and **leaf-wise (best-first) tree expansion**.

#### Key Features:
- Replaces sorting with discrete histogram bins (typically 256 bins).
- GOSS (Gradient-based One-Side Sampling) speeds up computation while retaining accuracy.
- Handles massive numerical and categorical datasets seamlessly.


```python
# Step 1: Install LightGBM if needed & Import Libraries
# !pip install lightgbm

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

try:
    import lightgbm as lgb
    print("LightGBM version:", lgb.__version__)
except ImportError:
    print("LightGBM not installed. Run: pip install lightgbm")
```

```python
# Step 2: Prepare Housing Data
housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(
    housing.data[:2000], housing.target[:2000], test_size=0.25, random_state=42
)
```

```python
# Step 3: Train LightGBM Regressor
try:
    lgb_reg = lgb.LGBMRegressor(
        n_estimators=100,
        learning_rate=0.05,
        num_leaves=31,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    )
    lgb_reg.fit(X_train, y_train)

    y_pred = lgb_reg.predict(X_test)
    print(f"LightGBM Test R²: {r2_score(y_test, y_pred):.4f}")
    print(f"LightGBM Test RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
except NameError:
    print("LightGBM not available.")
```

```python
# Step 4: Actual vs Predicted Plot
try:
    plt.figure(figsize=(7, 6))
    plt.scatter(y_test, y_pred, alpha=0.6, color='coral', edgecolors='k')
    plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2)
    plt.xlabel("Actual Values")
    plt.ylabel("Predicted Values")
    plt.title("LightGBM Regressor: Actual vs Predicted")
    plt.grid(True)
    plt.show()
except Exception as e:
    print(e)
```

### Summary
- **Strengths**: Extremely rapid training times, scalable to millions of rows, low memory footprint.
- **Weaknesses**: Leaf-wise tree growth can overfit small datasets if `num_leaves` is not restricted.
