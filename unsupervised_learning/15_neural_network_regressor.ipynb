# 15. Neural Network Regressor (Multi-Layer Perceptron)
### Learns Complex Nonlinear Relationships

An **MLP Regressor** is a feedforward artificial neural network trained via backpropagation to model complex non-linear mappings from inputs to a continuous output value.

#### Key Concepts:
1. **Network Architecture**:
   Input Layer $\rightarrow$ Multiple Hidden Layers with Non-linear Activations (ReLU, GELU) $\rightarrow$ Linear Output Layer ($y = W^{(out)} h + b$).
2. **Loss Function**: Mean Squared Error (MSE).
3. **Weight Regularization & Optimization**:
   $L_2$ weight decay (`alpha`) and Adam optimizer with adaptive learning rates.


```python
# Step 1: Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Step 2: Load and Scale Housing Data
housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(
    housing.data[:2000], housing.target[:2000], test_size=0.25, random_state=42
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Step 3: Train MLP Regressor
mlp_reg = MLPRegressor(
    hidden_layer_sizes=(64, 32),
    activation='relu',
    solver='adam',
    alpha=0.01,
    max_iter=500,
    random_state=42,
    early_stopping=True,
    n_iter_no_change=15
)
mlp_reg.fit(X_train_scaled, y_train)

y_pred = mlp_reg.predict(X_test_scaled)
print(f"MLP Regressor Test R²: {r2_score(y_test, y_pred):.4f}")
print(f"MLP Regressor Test RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
print(f"Iterations to converge: {mlp_reg.n_iter_}")
```

```python
# Step 4: Loss Curve Progression
plt.figure(figsize=(8, 4))
plt.plot(mlp_reg.loss_curve_, color='crimson', lw=2)
plt.xlabel("Epochs")
plt.ylabel("MSE Loss")
plt.title("MLP Regressor Training Loss Curve")
plt.grid(True)
plt.show()
```

### Summary
- **Strengths**: Universal approximator capable of learning arbitrarily complex non-linear functions, scalable to multimodal architectures.
- **Weaknesses**: Requires careful feature scaling and hyperparameter tuning; susceptible to local minima and overfitting on small datasets.
